```python
import csv
from shutil import copyfile
from pycocotools.coco import COCO
from tqdm import tqdm
```

```python
import os
import urllib.request
import zipfile

# Create data directory
os.makedirs("./data", exist_ok=True)

# Download and extract train images
train_url = "http://images.cocodataset.org/zips/train2014.zip"
train_zip = "./data/train2014.zip"

print("Downloading train2014 images (approx. 13GB)...")
urllib.request.urlretrieve(train_url, train_zip)
print("Extracting train images...")
with zipfile.ZipFile(train_zip, "r") as zip_ref:
  zip_ref.extractall("./data/")
os.remove(train_zip)
print("Train images ready!")
```

```python
# Download and extract validation images
val_url = "http://images.cocodataset.org/zips/val2014.zip"
val_zip = "./data/val2014.zip"

print("Downloading val2014 images (approx. 6GB)...")
urllib.request.urlretrieve(val_url, val_zip)
print("Extracting validation images...")
with zipfile.ZipFile(val_zip, "r") as zip_ref:
  zip_ref.extractall("./data/")
os.remove(val_zip)
print("Validation images ready!")
```

```python
coco = COCO('./data/annotations/captions_train2014.json')
```

```python
#get ids of training images
with open('TrainImageIds.csv', 'r') as f:
    reader = csv.reader(f)
    trainIds = list(reader)
    
trainIds = [int(i) for i in trainIds[0]]
print(len(trainIds))
```

```python
for img_id in tqdm(trainIds):
  path = coco.loadImgs(img_id)[0]["file_name"]
  # Updated local source path for Windows
  src_path = os.path.join("./data/train2014", path)
  dst_path = os.path.join("./data/images/train", path)
  copyfile(src_path, dst_path)
```
